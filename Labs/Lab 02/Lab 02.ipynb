```python
def bubble_sort(ar):
    n = len(ar)
    for i in range(n-1):
        swapped = False
        for j in range(n-i-1):
            if ar[j] > ar[j+1]:
                ar[j], ar[j+1] = ar[j+1], ar[j]
                swapped = True
        if not swapped:
            break
            
    return ar

def selection_sort(ar):
    for i in range(len(ar)):
        x = i
        for j in range(i, len(ar)):
            if ar[x] > ar[j]:
                x = j
            ar[i], ar[x] = ar[x], ar[i]
    return ar

def insertion_sort(ar):
    for i in range(1, len(ar)):
        key = ar[i]
        j = i - 1

        while j >= 0 and ar[j] > key:
            ar[j+1] = ar[j]
            j -= 1
        ar[j+1] = key
    return ar

def merge(left, right):
    result = []
    i = j = 0

    while i < len(left) and j < len(right):
        if left[i] < right[j]:
            result.append(left[i])
            i += 1
        else:
            result.append(right[j])
            j += 1
    result.extend(left[i:])
    result.extend(right[j:])

    return result
    
def merge_sort(ar):
    if len(ar) <= 1:
        return ar

    mid = len(ar) // 2
    left = merge_sort(ar[:mid])
    right = merge_sort(ar[mid:])

    return merge(left, right)

def quick_sort(ar):
    if len(ar) <= 1:
        return ar

    pivot = ar[-1]
    left = []
    right = []
    for i in range(len(ar)-1):
        if pivot > ar[i]:
            left.append(ar[i])
        else:
            right.append(ar[i])

    return quick_sort(left) + [pivot] + quick_sort(right)
    
        
print(bubble_sort(['P','Y','T','H','O','N']))
print(selection_sort(['P','Y','T','H','O','N']))
print(insertion_sort(['P','Y','T','H','O','N']))
print(merge_sort(['P','Y','T','H','O','N']))
print(quick_sort(['P','Y','T','H','O','N']))
```

```python
def statistics(ar):
    n = len(ar)

    #Mean
    mean = sum(ar) / n
    #Median
    xr = sorted(ar)
    if n % 2 == 1:
        median = xr[n // 2]
    else:
        median = (xr[n//2 - 1] + xr[n//2]) / 2
    #Mode
    freq = {}
    for num in ar:
        freq[num] = freq.get(num, 0) + 1
    max_freq = max(freq.values())
    mode = [k for k, v in freq.items() if v == max_freq]

    return mean, median, mode

data = [1, 2, 2, 3, 3, 4]
mean, median, mode = statistics(data)
print("Mean:", mean)
print("Median:", median)
print("Mode:", mode)
```

```python
def dup(ar):
    ar_dict = {}
    for num in ar:
        ar_dict[num] = ar_dict.get(num, 0) + 1
    return [k for k, v in ar_dict.items() if v > 1]
    
ar = []
ar = list(map(int, input("Enter the numbers to find duplicates: ").split()))

print(dup(ar))
```

```python
ar = list(range(100000, 300001))
def search(ar, key):
    for i in ar:
        if ar == key:
            return ar
    return None

import time
def timer(key):
    start = time.time()
    search(ar, key)
    end = time.time()
    return end - start

print("Front (100000):", timer(100000))
print("Middle (200000):", timer(200000))
print("Back (300000):", timer(300000))
print("Not in list (3):", timer(3))
```
